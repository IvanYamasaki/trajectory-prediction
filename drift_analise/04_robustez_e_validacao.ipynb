```python
import os, sys
from pathlib import Path
ROOT = Path.cwd().resolve()
while ROOT != ROOT.parent and not (ROOT / 'requirements.txt').exists():
    ROOT = ROOT.parent
os.chdir(ROOT)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
from drift_analise.drift_output_paths import CH01, CH02, CH03, CH04, COVARIATE_SHIFT_OUT, ensure_chapter_dirs
from drift_analise.notebook_utils import display_pdf, display_png, display_sorted_glob, save_figure
ensure_chapter_dirs()

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

IN_DIR = CH04
OUT_DIR = CH04
OUT_DIR.mkdir(parents=True, exist_ok=True)

PLOT_STYLE = {
    "font.family": "serif", "font.size": 12,
    "axes.titlesize": 13, "axes.labelsize": 12,
    "legend.fontsize": 10, "xtick.labelsize": 10, "ytick.labelsize": 10,
    "mathtext.fontset": "cm", "lines.linewidth": 1.8,
    "axes.grid": True, "grid.alpha": 0.25,
    "figure.autolayout": True,
}
plt.rcParams.update(PLOT_STYLE)


def T(pt: str, en: str, lang: str) -> str:
    return en if lang == "en" else pt


def plot_bic_curve(lang: str) -> None:
    csv = IN_DIR / "ruptures_bic_curve.csv"
    if not csv.exists():
        print(f"[warn] {csv} nao encontrado; pulando.")
        return
    df = pd.read_csv(csv)
    fig, axes = plt.subplots(1, 2, figsize=(11, 4.0))
    ax = axes[0]
    ax.plot(df["K"], df["m_bic"], "o-", color="#2A6FDB",
            label=T("mBIC (Schwarz)", "mBIC (Schwarz)", lang))
    K_star = int(df.loc[df["m_bic"].idxmin(), "K"])
    ax.axvline(K_star, color="#D1495B", lw=1.5, ls="--",
               label=T(f"$K^*={K_star}$ (ótimo BIC)", f"$K^*={K_star}$ (BIC-optimal)", lang))
    ax.set_xlabel(T(r"$K$ (número de breakpoints)", r"$K$ (number of breakpoints)", lang))
    ax.set_ylabel("mBIC")
    ax.set_title(T(r"Critério BIC modificado vs $K$", r"Modified BIC criterion vs $K$", lang))
    ax.legend(loc="lower right", framealpha=0.92)
    ax = axes[1]
    df2 = df.dropna(subset=["delta_cost"])
    ax.bar(df2["K"], df2["delta_cost"], color="#888", alpha=0.85,
           label=T(r"$\Delta\,\mathrm{custo}$ por bp adicional",
                   r"$\Delta\,\mathrm{cost}$ per added bp", lang))
    Tt = int(df["T"].iloc[0])
    bic_pen = -3 * np.log(Tt)
    ax.axhline(bic_pen, color="#D1495B", lw=1.4, ls="--",
               label=T(f"penalidade BIC: $-3\\log T = {bic_pen:.2f}$",
                       f"BIC penalty: $-3\\log T = {bic_pen:.2f}$", lang))
    ax.set_xlabel(T(r"$K$ (breakpoint adicionado)", r"$K$ (added breakpoint)", lang))
    ax.set_ylabel(T(r"$\Delta\,\mathrm{custo}$", r"$\Delta\,\mathrm{cost}$", lang))
    ax.set_title(T(r"Ganho marginal vs penalidade BIC", r"Marginal gain vs BIC penalty", lang))
    ax.legend(loc="upper right", framealpha=0.92)
    out = OUT_DIR / f"8_pelt_bic_curve_{lang}.pdf"
    saved = save_figure(fig, out, dpi=150)
    plt.close(fig)
    for p in saved:
        print(f"  -> {p}")


def plot_gamma_sensitivity(lang: str) -> None:
    csv = IN_DIR / "ruptures_gamma_sensitivity.csv"
    if not csv.exists():
        print(f"[warn] {csv} nao encontrado; pulando.")
        return
    df = pd.read_csv(csv)
    pivot = df.pivot(index="gamma_factor", columns="pen", values="n_bps")
    pivot = pivot.sort_index()
    fig, ax = plt.subplots(figsize=(8.5, 3.6))
    im = ax.imshow(pivot.values, cmap="YlGnBu", aspect="auto",
                   vmin=0, vmax=int(pivot.values.max()))
    ax.set_xticks(range(len(pivot.columns)))
    ax.set_xticklabels([str(c) for c in pivot.columns])
    ax.set_yticks(range(len(pivot.index)))
    ax.set_yticklabels([f"{g:.2g}" for g in pivot.index])
    for i in range(pivot.shape[0]):
        for j in range(pivot.shape[1]):
            v = pivot.values[i, j]
            color = "white" if v >= 2 else "black"
            ax.text(j, i, str(int(v)), ha="center", va="center",
                    color=color, fontsize=9)
    ax.set_xlabel(T(r"penalidade \texttt{pen}", r"penalty \texttt{pen}", lang))
    ax.set_ylabel(T(r"$\gamma$ (fator $\times$ default)", r"$\gamma$ (factor $\times$ default)", lang))
    ax.set_title(T(r"$n_\mathrm{bps}$ por $(\gamma, \mathrm{pen})$ no Pelt-RBF",
                   r"$n_\mathrm{bps}$ by $(\gamma, \mathrm{pen})$ in Pelt-RBF", lang))
    cbar = plt.colorbar(im, ax=ax, shrink=0.85)
    cbar.set_label(r"$n_\mathrm{bps}$")
    ax.grid(False)
    out = OUT_DIR / f"8_pelt_gamma_sensitivity_{lang}.pdf"
    saved = save_figure(fig, out, dpi=150)
    plt.close(fig)
    for p in saved:
        print(f"  -> {p}")


for lang in ["pt", "en"]:
    plot_bic_curve(lang)
    plot_gamma_sensitivity(lang)
print("[ok] figuras BIC e gamma sensitivity geradas.")
```
